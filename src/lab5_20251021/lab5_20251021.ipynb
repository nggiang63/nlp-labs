```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
import pandas as pd
```

```python
texts = [
    "This movie is fantastic and I love it!",
    "I hate this film, it's terrible.",
    "The acting was superb, a truly great experience.",
    "What a waste of time, absolutely boring.",
    "Highly recommend this, a masterpiece.",
    "Could not finish watching, so bad."
]

labels = [1, 0, 1, 0, 1, 0]  # 1 = positive, 0 = negative
```

```python
# CountVectorizer
vectorizer1 = CountVectorizer(stop_words='english')

X = vectorizer1.fit_transform(texts)

df = pd.DataFrame(X.toarray(), columns=vectorizer1.get_feature_names_out())
df['label'] = labels

df.head()
```

```python
# TF-IDF
vectorizer2 = TfidfVectorizer(stop_words='english')

X = vectorizer2.fit_transform(texts)

df = pd.DataFrame(X.toarray(), columns=vectorizer2.get_feature_names_out())
df['label'] = labels

df.head()
```
